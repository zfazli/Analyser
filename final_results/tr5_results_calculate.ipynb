```python
import torch
import re

datasets = [   
    "trace_e5" ,
    "trace_e5_1",
    "trace_e5_2",
    "trace_e5_3",
    "trace_e5_75",
    "trace_e5_50",
    "trace_e5_25",
    "trace_e5_1_random",
    "trace_e5_2_random",
    "trace_e5_3_random",
    "trace_e5_balanced",
    ]
name = 'TRACE_E5'

def calculate (tp , fp , tn , fn ):
    eps = 1e-10
    precision = tp/(tp+fp+eps)
    recall = tp/(tp+fn+eps)
    fscore = 2*precision*recall/(precision+recall+eps)

    print('tp,fp,tn,fn')
    print(tp,fp,tn,fn)
    print('Precision: ', precision)
    print('Recall: ', recall)
    print('F-Score: ', fscore)

def result (name , datasets , method):
  for dataset in datasets:
    print(f'########### {dataset} ###########')
    path = f"./results/{name}/{method}/{dataset}/results/results.pth"
    data = torch.load(path, map_location="cpu")

    tp = set()
    fp =  set()
    tn =  set()
    fn =  set()
    for  key in data.keys():
        if data[key]["y_hat"]==1:
            if data[key]["y_true"]==1:
                tp.add(key)
            else:
                fp.add(key)
        else :
            if data[key]["y_true"]==1:
                fn.add(key)
            else:
                tn.add(key)
    calculate(len(tp) , len(fp) , len(tn) , len(fn))

```

```python
method = 'flash'
result(name,datasets , method)
```

```python
method = 'velox'
result(name,datasets , method)
```

```python
method = 'orthrus'
result(name ,datasets, method)
```
